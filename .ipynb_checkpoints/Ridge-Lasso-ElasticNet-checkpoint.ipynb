Algerian Forest Fire Dataset

```python
#Import libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
#Load Dataset
df = pd.read_csv('Algerian_forest_fires_dataset_UPDATE.csv', header=1)
```

```python
df
```

```python
df.info()
```

# Data Cleaning

```python
#Check missing values
df[df.isnull().any(axis=1)]
```

```python
#Devide dataset based on region
```

```python
# Bejaia Region Dataset = 0
# Sidi-Bel Abbes Region Dataset = 1
df.loc[:122, "Region"] = 0
df.loc[122:, "Region"] = 1
```

```python
df.info()
```

```python
df[['Region']] = df[['Region']].astype(int)
```

```python
df.tail()
```

```python
# Remove null values
df.isnull().sum()
```

```python
df = df.dropna().reset_index(drop = True)
```

```python
df.tail()
```

```python
df.isnull().sum()
```

```python
df.iloc[[122]]
```

```python
df = df.drop(122).reset_index(drop=True)
```

```python
df.iloc[[122]]
```

```python
#Fix spaces in column name 
df.columns = df.columns.str.strip()
```

```python
df.columns
```

```python
#Change the requied column types to Int
```

```python
df[['month', 'day', 'year', 'Temperature', 'RH', 'Ws']] = df[['month', 'day', 'year', 'Temperature', 'RH', 'Ws']].astype(int)
```

```python
#Change the requied column types to Float

objects = [features for features in df.columns if df[features].dtypes=='O']
```

```python
for i in objects:
    if i != 'Classes':
        df[i] = df[i].astype(float)
```

```python
df.info()
```

```python
# Save the cleaned dataset
```

```python
df.to_csv('Cleaned_Dataset.csv', index=False)
```

# Exploratory Data Analysis

```python
#Drop day, month and year
df_copy = df.drop(['day', 'month', 'year'], axis=1)
```

```python
df_copy.head()
```

```python
# Do the encoding of categories in dataset 
```

```python
df_copy['Classes'] = np.where(df_copy['Classes'].str.contains('not fire'),0,1)
```

```python
df_copy.head()
```

Visualization

```python
#Plot density plot for all features
```

```python
plt.style.use('ggplot')
df_copy.hist(bins=50, figsize=(20, 15))
plt.show()
```

```python
#Percentage for piechar
```

```python
percentage = df_copy['Classes'].value_counts(normalize=True)*100
```

```python
 #Pie char
```

```python
classLabel = ['Fire', 'Not Fire']
plt.figure(figsize=(10, 7))
plt.pie(percentage, labels= classLabel, autopct='%1.1f%%')
plt.title("Pie chart")
plt.show()
```
